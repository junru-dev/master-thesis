```python
import os
import json
import re
import torch
import torchvision.transforms as T
import pandas as pd
from PIL import Image
from torchvision.transforms.functional import InterpolationMode
from tqdm import tqdm

IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD = (0.229, 0.224, 0.225)

# ===============================
# InternVL 官方图像预处理（动态切块）
# ===============================
def build_transform(input_size):
    return T.Compose([
        T.Lambda(lambda img: img.convert('RGB') if img.mode != 'RGB' else img),
        T.Resize((input_size, input_size), interpolation=InterpolationMode.BICUBIC),
        T.ToTensor(),
        T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
    ])

def find_closest_aspect_ratio(aspect_ratio, target_ratios, width, height, image_size):
    best_ratio_diff = float('inf')
    best_ratio = (1, 1)
    area = width * height
    for ratio in target_ratios:
        target_aspect_ratio = ratio[0] / ratio[1]
        ratio_diff = abs(aspect_ratio - target_aspect_ratio)
        if ratio_diff < best_ratio_diff:
            best_ratio_diff = ratio_diff
            best_ratio = ratio
        elif ratio_diff == best_ratio_diff:
            if area > 0.5 * image_size * image_size * ratio[0] * ratio[1]:
                best_ratio = ratio
    return best_ratio

def dynamic_preprocess(image, min_num=1, max_num=6, image_size=448, use_thumbnail=True):
    orig_width, orig_height = image.size
    aspect_ratio = orig_width / orig_height

    target_ratios = set(
        (i, j) for n in range(min_num, max_num + 1)
        for i in range(1, n + 1) for j in range(1, n + 1)
        if i * j <= max_num and i * j >= min_num
    )
    target_ratios = sorted(target_ratios, key=lambda x: x[0] * x[1])

    target_aspect_ratio = find_closest_aspect_ratio(
        aspect_ratio, target_ratios, orig_width, orig_height, image_size)

    target_width = image_size * target_aspect_ratio[0]
    target_height = image_size * target_aspect_ratio[1]
    blocks = target_aspect_ratio[0] * target_aspect_ratio[1]

    resized_img = image.resize((target_width, target_height))
    processed_images = []
    for i in range(blocks):
        box = (
            (i % (target_width // image_size)) * image_size,
            (i // (target_width // image_size)) * image_size,
            ((i % (target_width // image_size)) + 1) * image_size,
            ((i // (target_width // image_size)) + 1) * image_size
        )
        processed_images.append(resized_img.crop(box))

    if use_thumbnail and len(processed_images) != 1:
        processed_images.append(image.resize((image_size, image_size)))
    return processed_images

def load_image(image_path, input_size=448, max_num=6):
    image = Image.open(image_path).convert('RGB')
    transform = build_transform(input_size)
    images = dynamic_preprocess(image, image_size=input_size, use_thumbnail=True, max_num=max_num)
    pixel_values = torch.stack([transform(img) for img in images])
    return pixel_values

# ===============================
# 路径配置
# ===============================
test_image_dir = "/content/drive/MyDrive/MyThesis2026/data/cindy/images/test"
test_csv = "/content/drive/MyDrive/MyThesis2026/data/cindy/data/test.csv"
output_json = "/content/drive/MyDrive/InternVL25_8B_Misogyny_ZeroShot_pred.json"

test_df = pd.read_csv(test_csv)

prompt_text = """You are an expert in classifying harmful memes in the Chinese language. Your objective is to assess whether a meme is misogynistic or not.

Input: [Meme]
Image: [See attached image]
Text embedded: [Read from the image above]

Follow the steps below:
Step 1: Analyze the input meme by assessing both the image and the embedded text to determine whether it contains misogynistic content.
Step 2: If the meme contains any negative, insulting, stereotyping, or degrading reference to women, output Misogyny.
Step 3: If the meme does not contain any misogynistic content, output Non_Misogyny.

Output:
Your output should strictly follow the format:
Class labels: Misogyny or Non_Misogyny
Thought: Give your reason here"""

generation_config = dict(max_new_tokens=200, do_sample=False)

def parse_label(raw):
    m = re.search(r"Class labels?:\**\s*(Non_Misogyny|Misogyny)", raw, re.IGNORECASE)
    if m:
        matched = m.group(1).lower()
        return "Non_Misogyny" if matched == "non_misogyny" else "Misogyny"
    if any(w in raw.lower() for w in ["unable", "cannot", "can't", "sorry", "i'm not able"]):
        return "REFUSAL"
    if re.search(r"non[_\s-]?misogyn", raw, re.IGNORECASE):
        return "Non_Misogyny"
    if re.search(r"misogyn", raw, re.IGNORECASE):
        return "Misogyny"
    return "UNKNOWN"

# ===============================
# 断点续跑
# ===============================
if os.path.exists(output_json):
    with open(output_json, "r", encoding="utf-8") as f:
        predictions = json.load(f)
    predictions = [p for p in predictions if p["predicted_label"] != "ERROR"]
    done_images = {p["image_name"] for p in predictions}
    print(f"发现已有成功结果 {len(done_images)} 条，从断点继续...")
else:
    predictions = []
    done_images = set()
    print("没有已有结果，从头开始...")

remaining_df = test_df[~test_df["filename"].isin(done_images)]
print(f"剩余待处理: {len(remaining_df)} 张")

# ===============================
# 主循环
# ===============================
for _, row in tqdm(remaining_df.iterrows(), total=len(remaining_df), desc="推理进度"):
    img_name = row["filename"]
    img_path = os.path.join(test_image_dir, img_name)

    if not os.path.exists(img_path):
        print(f"⚠️ 图片不存在: {img_name}")
        continue

    try:
        torch.cuda.empty_cache()
        pixel_values = load_image(img_path).to(torch.float16).cuda()

        response = model.chat(tokenizer, pixel_values, prompt_text, generation_config)
        label = parse_label(response)

        predictions.append({
            "image_name": img_name,
            "predicted_label": label,
            "raw_output": response,
            "true_label": int(row["label"])
        })

        gt = 'Misogyny' if row['label'] == 1 else 'Non_Misogyny'
        print(f"✅ {img_name} -> {label} (真实: {gt})")

    except Exception as e:
        print(f"❌ {img_name} 出错: {e}")
        predictions.append({
            "image_name": img_name,
            "predicted_label": "ERROR",
            "raw_output": str(e),
            "true_label": int(row["label"])
        })

    with open(output_json, "w", encoding="utf-8") as f:
        json.dump(predictions, f, ensure_ascii=False, indent=2)
    torch.cuda.empty_cache()

# ===============================
# 统计
# ===============================
print(f"\n完成！共 {len(predictions)} 条结果已保存")
labels = [p["predicted_label"] for p in predictions]
for lbl in sorted(set(labels)):
    print(f"  {lbl}: {labels.count(lbl)}")
```

Calculate zero-shot result

```python
import json
import pandas as pd
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)

json_path = "/content/drive/MyDrive/InternVL25_8B_Misogyny_ZeroShot_pred.json"

with open(json_path, "r", encoding="utf-8") as f:
    predictions = json.load(f)

df = pd.DataFrame(predictions)
print(f"总条数: {len(df)}")

print("\n预测标签分布:")
print(df["predicted_label"].value_counts())

# ---- 剔除异常样本，但单独统计 ----
BAD = ["ERROR", "UNKNOWN", "REFUSAL"]
bad_df = df[df["predicted_label"].isin(BAD)]
clean_df = df[~df["predicted_label"].isin(BAD)].copy()

print(f"\n有效样本: {len(clean_df)} / {len(df)}")
if len(bad_df) > 0:
    print("异常样本:")
    print(bad_df["predicted_label"].value_counts())
    print(f"拒答率: {(df['predicted_label']=='REFUSAL').mean()*100:.2f}%")

# ---- 标签统一 ----
clean_df["y_true"] = clean_df["true_label"].map({0: "Non_Misogyny", 1: "Misogyny"})
clean_df["y_pred"] = clean_df["predicted_label"]

y_true = clean_df["y_true"]
y_pred = clean_df["y_pred"]
LABELS = ["Non_Misogyny", "Misogyny"]

# ---- 七个指标 ----
ACC = accuracy_score(y_true, y_pred)
MP  = precision_score(y_true, y_pred, average="macro",    labels=LABELS, zero_division=0)
MR  = recall_score(y_true, y_pred,    average="macro",    labels=LABELS, zero_division=0)
MF1 = f1_score(y_true, y_pred,        average="macro",    labels=LABELS, zero_division=0)
WP  = precision_score(y_true, y_pred, average="weighted", labels=LABELS, zero_division=0)
WR  = recall_score(y_true, y_pred,    average="weighted", labels=LABELS, zero_division=0)
WF1 = f1_score(y_true, y_pred,        average="weighted", labels=LABELS, zero_division=0)

print("\n" + "="*49)
print("InternVL2.5-8B | Misogyny | Zero-shot")
print("="*49)
print(f"ACC  : {ACC:.4f}")
print(f"MP   : {MP:.4f}")
print(f"MR   : {MR:.4f}")
print(f"MF1  : {MF1:.4f}   ← 主指标")
print(f"WP   : {WP:.4f}")
print(f"WR   : {WR:.4f}")
print(f"WF1  : {WF1:.4f}")
print("="*49)

print("\nConfusion Matrix (行=真实, 列=预测):")
print(f"labels: {LABELS}")
print(confusion_matrix(y_true, y_pred, labels=LABELS))

print("\nClassification Report:")
print(classification_report(y_true, y_pred, labels=LABELS, zero_division=0))
```

```python
import os
import json
import re
import torch
import numpy as np
import torchvision.transforms as T
import pandas as pd
from PIL import Image
from torchvision.transforms.functional import InterpolationMode
from tqdm import tqdm
from transformers import CLIPProcessor, CLIPModel

IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD = (0.229, 0.224, 0.225)

# ===============================
# InternVL 官方图像预处理（动态切块）
# ===============================
def build_transform(input_size):
    return T.Compose([
        T.Lambda(lambda img: img.convert('RGB') if img.mode != 'RGB' else img),
        T.Resize((input_size, input_size), interpolation=InterpolationMode.BICUBIC),
        T.ToTensor(),
        T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
    ])

def find_closest_aspect_ratio(aspect_ratio, target_ratios, width, height, image_size):
    best_ratio_diff = float('inf')
    best_ratio = (1, 1)
    area = width * height
    for ratio in target_ratios:
        target_aspect_ratio = ratio[0] / ratio[1]
        ratio_diff = abs(aspect_ratio - target_aspect_ratio)
        if ratio_diff < best_ratio_diff:
            best_ratio_diff = ratio_diff
            best_ratio = ratio
        elif ratio_diff == best_ratio_diff:
            if area > 0.5 * image_size * image_size * ratio[0] * ratio[1]:
                best_ratio = ratio
    return best_ratio

def dynamic_preprocess(image, min_num=1, max_num=6, image_size=448, use_thumbnail=True):
    orig_width, orig_height = image.size
    aspect_ratio = orig_width / orig_height

    target_ratios = set(
        (i, j) for n in range(min_num, max_num + 1)
        for i in range(1, n + 1) for j in range(1, n + 1)
        if i * j <= max_num and i * j >= min_num
    )
    target_ratios = sorted(target_ratios, key=lambda x: x[0] * x[1])

    target_aspect_ratio = find_closest_aspect_ratio(
        aspect_ratio, target_ratios, orig_width, orig_height, image_size)

    target_width = image_size * target_aspect_ratio[0]
    target_height = image_size * target_aspect_ratio[1]
    blocks = target_aspect_ratio[0] * target_aspect_ratio[1]

    resized_img = image.resize((target_width, target_height))
    processed_images = []
    for i in range(blocks):
        box = (
            (i % (target_width // image_size)) * image_size,
            (i // (target_width // image_size)) * image_size,
            ((i % (target_width // image_size)) + 1) * image_size,
            ((i // (target_width // image_size)) + 1) * image_size
        )
        processed_images.append(resized_img.crop(box))

    if use_thumbnail and len(processed_images) != 1:
        processed_images.append(image.resize((image_size, image_size)))
    return processed_images

def load_image(image_path, input_size=448, max_num=6):
    image = Image.open(image_path).convert('RGB')
    transform = build_transform(input_size)
    images = dynamic_preprocess(image, image_size=input_size, use_thumbnail=True, max_num=max_num)
    pixel_values = torch.stack([transform(img) for img in images])
    return pixel_values

# ===============================
# 加载 CLIP + misogyny 训练集 embeddings（RAG 检索用）
# ===============================
print("Loading CLIP...")
clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
clip_model.eval()

train_embeddings = np.load("/content/drive/MyDrive/misogyny_train_embeddings.npy")
with open("/content/drive/MyDrive/misogyny_train_meta.json", "r") as f:
    train_meta = json.load(f)

train_labels = train_meta["labels"]       # 0 / 1 整数
train_filenames = train_meta["filenames"]
print(f"训练集 embeddings 加载完成，共 {len(train_embeddings)} 条")

def get_clip_embedding(image_path):
    image = Image.open(image_path).convert("RGB")
    inputs = clip_processor(images=image, return_tensors="pt")
    with torch.no_grad():
        outputs = clip_model.vision_model(**inputs)
        emb = outputs.pooler_output
        emb = emb / emb.norm(dim=-1, keepdim=True)
    return emb.squeeze().numpy()

def retrieve_examples(test_emb):
    similarities = train_embeddings @ test_emb
    label_map = {1: "Misogyny", 0: "Non_Misogyny"}
    examples = {}
    for label_id, label_name in label_map.items():
        label_indices = [i for i, l in enumerate(train_labels) if l == label_id]
        label_sims = [(i, similarities[i]) for i in label_indices]
        label_sims.sort(key=lambda x: x[1], reverse=True)
        examples[label_name] = label_sims[0][0]
    return examples

def build_prompt(example_indices):
    examples_text = ""
    for i, (label, idx) in enumerate(example_indices.items()):
        examples_text += f"Example {i+1}: Class label: {label}\n"

    return f"""You are an expert in classifying harmful memes in the Chinese language. Your objective is to assess whether a meme is misogynistic or not.

Below are 2 reference examples with their correct labels retrieved from similar memes:

{examples_text}
Now classify the following meme:

Input: [Meme]
Image: [See attached image]
Text embedded: [Read from the image above]

Follow the steps below:
Step 1: Analyze the input meme by assessing both the image and the embedded text, using the provided examples as reference, to determine whether it contains misogynistic content.
Step 2: If the meme contains any negative, insulting, stereotyping, or degrading reference to women, output Misogyny.
Step 3: If the meme does not contain any misogynistic content, output Non_Misogyny.

Output:
Your output should strictly follow the format:
Class labels: Misogyny or Non_Misogyny
Thought: Give your reason here"""

# ===============================
# 路径配置
# ===============================
test_image_dir = "/content/drive/MyDrive/MyThesis2026/data/cindy/images/test"
test_csv = "/content/drive/MyDrive/MyThesis2026/data/cindy/data/test.csv"
output_json = "/content/drive/MyDrive/InternVL25_8B_Misogyny_FewShot_RAG_pred.json"

test_df = pd.read_csv(test_csv)

def parse_label(raw):
    m = re.search(r"Class labels?:\**\s*(Non_Misogyny|Misogyny)", raw, re.IGNORECASE)
    if m:
        matched = m.group(1).lower()
        return "Non_Misogyny" if matched == "non_misogyny" else "Misogyny"
    if any(w in raw.lower() for w in ["unable", "cannot", "can't", "sorry", "i'm not able"]):
        return "REFUSAL"
    if re.search(r"non[_\s-]?misogyn", raw, re.IGNORECASE):
        return "Non_Misogyny"
    if re.search(r"misogyn", raw, re.IGNORECASE):
        return "Misogyny"
    return "UNKNOWN"

generation_config = dict(max_new_tokens=200, do_sample=False)

# ===============================
# 断点续跑
# ===============================
if os.path.exists(output_json):
    with open(output_json, "r", encoding="utf-8") as f:
        predictions = json.load(f)
    predictions = [p for p in predictions if p["predicted_label"] != "ERROR"]
    done_images = {p["image_name"] for p in predictions}
    print(f"发现已有成功结果 {len(done_images)} 条，从断点继续...")
else:
    predictions = []
    done_images = set()
    print("没有已有结果，从头开始...")

remaining_df = test_df[~test_df["filename"].isin(done_images)]
print(f"剩余待处理: {len(remaining_df)} 张")

# ------ 小规模测试用，先跑通再删掉这行 ------
# remaining_df = remaining_df.head(5)
# ------------------------------------------

# ===============================
# 主循环
# ===============================
for _, row in tqdm(remaining_df.iterrows(), total=len(remaining_df), desc="推理进度"):
    img_name = row["filename"]
    img_path = os.path.join(test_image_dir, img_name)

    if not os.path.exists(img_path):
        print(f"⚠️ 图片不存在: {img_name}")
        continue

    try:
        torch.cuda.empty_cache()

        # RAG 检索
        test_emb = get_clip_embedding(img_path)
        example_indices = retrieve_examples(test_emb)
        prompt_text = build_prompt(example_indices)

        # InternVL 动态切块预处理
        # 改动：精度从 float16 改为 bfloat16，与模型加载时的 torch_dtype=torch.bfloat16 保持一致
        pixel_values = load_image(img_path).to(torch.bfloat16).cuda()

        response = model.chat(tokenizer, pixel_values, prompt_text, generation_config)
        label = parse_label(response)

        predictions.append({
            "image_name": img_name,
            "predicted_label": label,
            "raw_output": response,
            "true_label": int(row["label"])
        })

        gt = 'Misogyny' if row['label'] == 1 else 'Non_Misogyny'
        print(f"✅ {img_name} -> {label} (真实: {gt})")

    except Exception as e:
        print(f"❌ {img_name} 出错: {e}")
        predictions.append({
            "image_name": img_name,
            "predicted_label": "ERROR",
            "raw_output": str(e),
            "true_label": int(row["label"])
        })

    with open(output_json, "w", encoding="utf-8") as f:
        json.dump(predictions, f, ensure_ascii=False, indent=2)
    torch.cuda.empty_cache()

# ===============================
# 统计
# ===============================
print(f"\n完成！共 {len(predictions)} 条结果已保存")
labels = [p["predicted_label"] for p in predictions]
for lbl in sorted(set(labels)):
    print(f"  {lbl}: {labels.count(lbl)}")
```

 Calculation of Few shot

```python
import json
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)

# ===============================
# 路径配置
# ===============================
pred_json = "/content/drive/MyDrive/InternVL25_8B_Misogyny_FewShot_RAG_pred.json"

with open(pred_json, "r", encoding="utf-8") as f:
    predictions = json.load(f)

print(f"总样本数: {len(predictions)}")

# ===============================
# 先分离 ERROR / UNKNOWN / REFUSAL
# ===============================
valid = []
error_count = 0
unknown_count = 0
refusal_count = 0

for p in predictions:
    label = p["predicted_label"]
    if label == "ERROR":
        error_count += 1
    elif label == "UNKNOWN":
        unknown_count += 1
    elif label == "REFUSAL":
        refusal_count += 1
    else:
        valid.append(p)

print(f"\n剔除统计：")
print(f"  ERROR: {error_count}")
print(f"  UNKNOWN: {unknown_count}")
print(f"  REFUSAL: {refusal_count}")
print(f"  参与评估的有效样本: {len(valid)}")

# ===============================
# 标签映射：字符串 -> 整数，与 true_label 对齐
# ===============================
label_map = {"Misogyny": 1, "Non_Misogyny": 0}

y_true = [p["true_label"] for p in valid]
y_pred = [label_map[p["predicted_label"]] for p in valid]

# ===============================
# 计算七个指标
# ===============================
acc = accuracy_score(y_true, y_pred)
macro_p = precision_score(y_true, y_pred, average="macro", zero_division=0)
macro_r = recall_score(y_true, y_pred, average="macro", zero_division=0)
macro_f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)
weighted_p = precision_score(y_true, y_pred, average="weighted", zero_division=0)
weighted_r = recall_score(y_true, y_pred, average="weighted", zero_division=0)
weighted_f1 = f1_score(y_true, y_pred, average="weighted", zero_division=0)

print("\n===== 评估结果 (InternVL2.5-8B Misogyny Few-shot RAG) =====")
print(f"Accuracy:      {acc:.4f}")
print(f"Macro-P:       {macro_p:.4f}")
print(f"Macro-R:       {macro_r:.4f}")
print(f"Macro-F1:      {macro_f1:.4f}")
print(f"Weighted-P:    {weighted_p:.4f}")
print(f"Weighted-R:    {weighted_r:.4f}")
print(f"Weighted-F1:   {weighted_f1:.4f}")

# ===============================
# 详细分类报告 + 混淆矩阵（方便查错）
# ===============================
target_names = ["Non_Misogyny", "Misogyny"]
print("\n===== 分类详情 =====")
print(classification_report(y_true, y_pred, target_names=target_names, zero_division=0))

print("===== 混淆矩阵 =====")
print("        预测:Non_Miso  预测:Miso")
cm = confusion_matrix(y_true, y_pred)
for i, row_name in enumerate(target_names):
    print(f"真实:{row_name:<12} {cm[i][0]:>10}  {cm[i][1]:>8}")
```
